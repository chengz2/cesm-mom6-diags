## Zonal means

If `Misc: zonal_mean_file` is set in `diag_config.yml`, each case minus the reference is shown: either another
case's `<casename>_zonal_means_????-????.nc`, or `self` for the case's own mean over model years
`--ref_year_start` to `--ref_year_end` (default 1-20), labeled with the reference's SNAME.
Otherwise, each case's own theta, salinity and sigma2 sections are shown.

```python
# figures written by mom6-tools_zonal_means, one per region with theta, salinity and sigma2 panels:
# the difference from the reference if Misc: zonal_mean_file is set, otherwise the case's own sections
regions = ['Circumpolar', 'Weddell', 'Indian', 'WestPacific', 'Ross', 'AmundsenBellingshausen']
kind = 'zonal_means_diff' if misc_config.get('zonal_mean_file') else 'zonal_means'
for i in range(len(casename)):
  for reg in regions:
    png_file = os.path.join(png_dir, 'zonal_means', '{}_{}_{}.png'.format(casename[i], kind, reg))
    if os.path.exists(png_file):
      display(Image(filename=png_file))
    else:
      print('Not found:', png_file)
```
